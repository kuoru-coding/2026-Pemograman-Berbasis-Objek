Praktikum 12.3

```python
from abc import ABC, abstractmethod

# --- PRAKTIKUM 3: Mendesain Kelas OOP ---
class Lokasi(ABC):
    def __init__(self, nama: str, latitude: float, longitude: float):
        self.nama = str(nama) if nama else "Tanpa Nama"
        try:
            self.latitude = float(latitude)
            self.longitude = float(longitude)
        except (ValueError, TypeError, SystemError):
            self.latitude, self.longitude = 0.0, 0.0

    def get_koordinat(self) -> tuple:
        return (self.latitude, self.longitude)

    @abstractmethod
    def get_info_popup(self) -> str:
        pass

    def __repr__(self) -> str:
        return f"{type(self).__name__}(nama='{self.nama}', lat={self.latitude:.4f}, lon={self.longitude:.4f})"

    def __str__(self) -> str:
        return f"{self.nama} [{type(self).__name__}]"

class TempatWisata(Lokasi):
    def __init__(self, nama: str, latitude: float, longitude: float, jenis: str, deskripsi: str):
        super().__init__(nama, latitude, longitude)
        self.jenis_wisata = str(jenis) if jenis else "Umum"
        self.deskripsi = str(deskripsi) if deskripsi else "Tidak ada deskripsi."

    def get_info_popup(self) -> str:
        return f"{self.nama}{self.jenis_wisata}{self.deskripsi}Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})"

class Kuliner(Lokasi):
    def __init__(self, nama: str, latitude: float, longitude: float, menu_andalan: str):
        super().__init__(nama, latitude, longitude)
        self.menu_andalan = str(menu_andalan) if menu_andalan else "Tidak diketahui"

    def get_info_popup(self) -> str:
        return f"{self.nama}KulinerMenu Andalan: {self.menu_andalan}Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})"

class TempatIbadah(Lokasi):
    def __init__(self, nama: str, latitude: float, longitude: float, agama: str = "Umum", deskripsi: str = ""):
        super().__init__(nama, latitude, longitude)
        self.agama = str(agama) if agama else "Umum"
        self.deskripsi = str(deskripsi) if deskripsi else "Tempat Ibadah"

    def get_info_popup(self) -> str:
        return f"{self.nama}Tempat Ibadah ({self.agama}){self.deskripsi}Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})"

print("--- Praktikum 3 Selesai: Kelas OOP berhasil didefinisikan ---")
```

Praktikum 12.4

```python
import pandas as pd

# --- PRAKTIKUM 4: Membuat Objek dari Data Pandas ---
def buat_objek_lokasi_dari_df(dataframe: pd.DataFrame) -> list:
    list_objek_lokasi = []

    if dataframe is None or dataframe.empty:
        print("DataFrame kosong atau None, tidak ada objek dibuat.")
        return list_objek_lokasi

    print("\nMembuat objek dari DataFrame...")
    for index, row in dataframe.iterrows():
        nama = row.get('Nama', None)
        lat = row.get('Latitude', None)
        lon = row.get('Longitude', None)
        tipe = row.get('Tipe', 'Lainnya')
        deskripsi = row.get('Deskripsi', '')

        if pd.isna(nama) or pd.isna(lat) or pd.isna(lon):
            continue

        objek = None
        try:
            if 'Wisata' in tipe or tipe == 'Landmark':
                objek = TempatWisata(nama, lat, lon, tipe, deskripsi)
            elif tipe == 'Kuliner':
                objek = Kuliner(nama, lat, lon, deskripsi)
            elif 'Ibadah' in tipe:
                agama_info = "Umum"
                if "Islam" in tipe: agama_info = "Islam"
                elif "Kristen" in tipe: agama_info = "Kristen"
                elif "Klenteng" in tipe: agama_info = "Tridharma"
                objek = TempatIbadah(nama, lat, lon, agama_info, deskripsi)

            if objek:
                list_objek_lokasi.append(objek)

        except Exception as e:
            print(f" -> GAGAL membuat objek untuk '{nama}' di baris {index}: {e}")

    print(f"Total {len(list_objek_lokasi)} objek lokasi berhasil dibuat dari {len(dataframe)} baris data.")
    return list_objek_lokasi

# Eksekusi Utama Praktikum 4
print("--- Memulai Praktikum 4: Membuat Objek dari Data Pandas ---")

# Memastikan df_lokasi dari Praktikum 2 sudah ada
try:
    list_semua_lokasi = buat_objek_lokasi_dari_df(df_lokasi)
    print("\n--- Daftar Objek Lokasi yang Berhasil Dibuat ---")
    if list_semua_lokasi:
        for idx, lok in enumerate(list_semua_lokasi):
            print(f" {idx+1}. {repr(lok)}")
    else:
        print("Tidak ada objek lokasi yang dibuat.")
except NameError:
    print("ERROR: Variabel 'df_lokasi' belum didefinisikan. Pastikan Cell Praktikum 2 sudah dijalankan!")

print("\n--- Praktikum 4 Selesai ---")
```

```python
import folium
import datetime
import os

FILE_LOG = "proses_peta.log"

# --- PRAKTIKUM 6: Fungsi Logging ---
def tulis_log(pesan: str, file_log: str = FILE_LOG):
    """Menulis pesan log ke file dengan timestamp, menggunakan mode append."""
    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    try:
        with open(file_log, 'a', encoding='utf-8') as f:
            f.write(f"[{timestamp}] {pesan}\n")
    except IOError as e:
        print(f"ERROR: Gagal menulis ke file log '{file_log}': {e}")

# --- PRAKTIKUM 6: Pembuatan Peta yang Diperbarui (dengan Logging) ---
def buat_peta_lokasi_folium_log(list_objek: list, file_output: str):
    nama_fungsi = "buat_peta_lokasi_folium"

    if not list_objek:
        pesan_log = f"[{nama_fungsi}] Gagal: Tidak ada data lokasi untuk dipetakan."
        print(pesan_log)
        tulis_log(pesan_log)
        return None

    print(f"\n[{nama_fungsi}] Memulai pembuatan peta dari {len(list_objek)} lokasi...")
    tulis_log(f"[{nama_fungsi}] Memulai pembuatan peta '{file_output}' dengan {len(list_objek)} lokasi.")

    try:
        lat_tengah = list_objek[0].latitude
        lon_tengah = list_objek[0].longitude
    except IndexError:
        lat_tengah, lon_tengah = -6.9929, 110.4200

    peta = folium.Map(location=[lat_tengah, lon_tengah], zoom_start=12)

    jumlah_marker = 0
    lokasi_dilewati = []

    for lok in list_objek:
        koordinat = lok.get_koordinat()
        if koordinat != (0.0, 0.0):
            info_popup_html = lok.get_info_popup()
            folium.Marker(
                location=koordinat,
                popup=folium.Popup(info_popup_html, max_width=300),
                tooltip=lok.nama
            ).add_to(peta)
            jumlah_marker += 1
        else:
            lokasi_dilewati.append(lok.nama)

    if lokasi_dilewati:
        pesan_lewat = f"[{nama_fungsi}] Melewati marker untuk: {', '.join(lokasi_dilewati)} (koordinat tidak valid)."
        print(f" -> Peringatan: {pesan_lewat}")
        tulis_log(pesan_lewat)

    try:
        peta.save(file_output)
        pesan_sukses = f"[{nama_fungsi}] Peta '{file_output}' berhasil dibuat dengan {jumlah_marker} marker."
        print(f"-> {pesan_sukses}")
        tulis_log(pesan_sukses)
    except Exception as e:
        pesan_error = f"[{nama_fungsi}] ERROR saat menyimpan peta '{file_output}': {type(e).__name__} - {e}"
        print(f"-> {pesan_error}")
        tulis_log(pesan_error)

    return peta

# Eksekusi Praktikum 6
print("--- Memulai Praktikum 6: File Handling Tambahan (Log) ---")

# (Opsional) Hapus log lama jika ingin mulai bersih saat di-run ulang
if os.path.exists(FILE_LOG):
    os.remove(FILE_LOG)
    print(f"File log lama '{FILE_LOG}' dihapus.")

# Kita buat peta kedua untuk demo logging
print("\nMenjalankan pembuatan peta lagi untuk demo log append...")
peta_dengan_log = buat_peta_lokasi_folium_log(list_semua_lokasi, "peta_kedua.html")

print(f"\nSilakan periksa isi file log '{FILE_LOG}' untuk melihat catatan proses.")

print("\n--- Isi dari proses_peta.log ---")
with open(FILE_LOG, 'r') as f:
    print(f.read())

print("--- Praktikum 6 Selesai ---")

# Tampilkan di Colab
peta_dengan_log
```

1.2 Definisi Semua OOP (termasuk kelas baru)

```python
import pandas as pd
import folium
import datetime
from abc import ABC, abstractmethod


# ── Kelas Dasar (Abstrak) ────────────────────────────────────────────────────
class Lokasi(ABC):
    """Kelas abstrak sebagai blueprint untuk semua jenis lokasi geografis."""

    def __init__(self, nama: str, latitude: float, longitude: float):
        self.nama = str(nama) if nama else "Tanpa Nama"
        try:
            self.latitude  = float(latitude)
            self.longitude = float(longitude)
        except (ValueError, TypeError):
            self.latitude  = 0.0
            self.longitude = 0.0

    def get_koordinat(self) -> tuple:
        return (self.latitude, self.longitude)

    @abstractmethod
    def get_info_popup(self) -> str:
        pass

    def __repr__(self) -> str:
        return (f"{type(self).__name__}(nama='{self.nama}', "
                f"lat={self.latitude:.4f}, lon={self.longitude:.4f})")

    def __str__(self) -> str:
        return f"{self.nama} [{type(self).__name__}]"


# ── Kelas Konkret (dari Praktikum) ──────────────────────────────────────────
class TempatWisata(Lokasi):
    def __init__(self, nama, latitude, longitude, jenis, deskripsi):
        super().__init__(nama, latitude, longitude)
        self.jenis_wisata = str(jenis)    if jenis    else "Umum"
        self.deskripsi    = str(deskripsi) if deskripsi else "Tidak ada deskripsi."

    def get_info_popup(self) -> str:
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>{self.jenis_wisata}</i><br><br>"
                f"{self.deskripsi}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


class Kuliner(Lokasi):
    def __init__(self, nama, latitude, longitude, menu_andalan):
        super().__init__(nama, latitude, longitude)
        self.menu_andalan = str(menu_andalan) if menu_andalan else "Tidak diketahui"

    def get_info_popup(self) -> str:
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>Kuliner</i><br><br>"
                f"Menu Andalan: {self.menu_andalan}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


class TempatIbadah(Lokasi):
    def __init__(self, nama, latitude, longitude, agama="Umum", deskripsi=""):
        super().__init__(nama, latitude, longitude)
        self.agama     = str(agama)     if agama     else "Umum"
        self.deskripsi = str(deskripsi) if deskripsi else "Tempat Ibadah"

    def get_info_popup(self) -> str:
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>Tempat Ibadah ({self.agama})</i><br><br>"
                f"{self.deskripsi}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


# ── Kelas Baru (Penugasan) ───────────────────────────────────────────────────
class Museum(Lokasi):
    """Kelas untuk merepresentasikan museum. Mewarisi dari Lokasi."""

    def __init__(self, nama: str, latitude: float, longitude: float,
                 koleksi_utama: str = "", deskripsi: str = ""):
        super().__init__(nama, latitude, longitude)
        self.koleksi_utama = str(koleksi_utama) if koleksi_utama else "Beragam koleksi"
        self.deskripsi     = str(deskripsi)     if deskripsi     else "Museum"

    def get_info_popup(self) -> str:
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>🏛️ Museum</i><br><br>"
                f"{self.deskripsi}<br><br>"
                f"<b>Koleksi Utama:</b> {self.koleksi_utama}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


class KantorPemerintahan(Lokasi):
    """Kelas untuk merepresentasikan kantor pemerintahan. Mewarisi dari Lokasi."""

    def __init__(self, nama: str, latitude: float, longitude: float,
                 instansi: str = "", deskripsi: str = ""):
        super().__init__(nama, latitude, longitude)
        self.instansi  = str(instansi)  if instansi  else "Instansi Pemerintah"
        self.deskripsi = str(deskripsi) if deskripsi else "Kantor Pemerintahan"

    def get_info_popup(self) -> str:
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>🏢 Kantor Pemerintahan</i><br><br>"
                f"{self.deskripsi}<br><br>"
                f"<b>Instansi:</b> {self.instansi}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


print("Semua kelas berhasil didefinisikan:")
print("  - Lokasi (abstrak)")
print("  - TempatWisata, Kuliner, TempatIbadah  (dari praktikum)")
print("  - Museum, KantorPemerintahan           (kelas baru – penugasan)")
```

2 Kustomisasi Marker & Kode utama

```python
def tulis_log(pesan: str, file_log: str = "proses_peta.log"):
    """Menulis pesan log ke file dengan timestamp (mode append)."""
    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    try:
        with open(file_log, 'a', encoding='utf-8') as f:
            f.write(f"[{timestamp}] {pesan}\n")
    except IOError as e:
        print(f"ERROR menulis log: {e}")


def baca_config_peta(file_config: str = "config_peta.txt"):
    """
    Membaca konfigurasi peta dari file teks.
    Format: baris 1 = latitude, baris 2 = longitude, baris 3 = zoom.
    Mengembalikan (lat, lon, zoom) atau nilai default Semarang jika error.
    """
    default = (-6.9929, 110.4200, 13)
    try:
        with open(file_config, 'r', encoding='utf-8') as f:
            baris = f.read().strip().splitlines()
        lat  = float(baris[0])
        lon  = float(baris[1])
        zoom = int(baris[2])
        print(f"Config peta dibaca: lat={lat}, lon={lon}, zoom={zoom}")
        return lat, lon, zoom
    except FileNotFoundError:
        print(f"Config '{file_config}' tidak ditemukan, menggunakan default.")
        return default
    except (ValueError, IndexError) as e:
        print(f"Error membaca config ({e}), menggunakan default.")
        return default


def dapatkan_ikon(lok: Lokasi) -> folium.Icon:
    """
    Mengembalikan folium.Icon berdasarkan tipe objek lokasi.
    Menggunakan isinstance() untuk memeriksa tipe objek.
    """
    if isinstance(lok, Museum):
        return folium.Icon(color='purple', icon='university', prefix='fa')
    elif isinstance(lok, KantorPemerintahan):
        return folium.Icon(color='gray',   icon='building',   prefix='fa')
    elif isinstance(lok, TempatWisata):
        return folium.Icon(color='blue',   icon='camera',     prefix='fa')
    elif isinstance(lok, Kuliner):
        return folium.Icon(color='red',    icon='cutlery',    prefix='fa')
    elif isinstance(lok, TempatIbadah):
        return folium.Icon(color='green',  icon='star',       prefix='fa')
    else:
        return folium.Icon(color='beige',  icon='map-marker', prefix='fa')


def buat_peta_lokasi_folium(list_objek: list,
                             file_output:  str = "peta_lokasi.html",
                             file_config:  str = "config_peta.txt",
                             file_log:     str = "proses_peta.log"):
    """
    Membuat peta Folium interaktif lengkap:
      - Membaca konfigurasi dari file teks
      - Marker dikustomisasi berdasarkan tipe objek
      - Menyimpan log setiap proses
    """
    if not list_objek:
        pesan = "Gagal: tidak ada data lokasi untuk dipetakan."
        print(pesan); tulis_log(pesan, file_log)
        return

    # 1. Baca konfigurasi peta dari file
    lat_tengah, lon_tengah, zoom = baca_config_peta(file_config)

    tulis_log(f"Memulai pembuatan peta '{file_output}' "
              f"dengan {len(list_objek)} lokasi.", file_log)

    # 2. Buat objek peta Folium
    peta = folium.Map(location=[lat_tengah, lon_tengah],
                      zoom_start=zoom,
                      tiles="OpenStreetMap")
    print(f"Peta dibuat berpusat di ({lat_tengah}, {lon_tengah}), zoom {zoom}")

    # 3. Tambahkan marker per objek dengan ikon/warna berbeda
    jumlah_marker  = 0
    lokasi_dilewati = []

    for lok in list_objek:
        koordinat = lok.get_koordinat()

        if koordinat == (0.0, 0.0):
            lokasi_dilewati.append(lok.nama)
            continue

        info_popup = lok.get_info_popup()
        ikon       = dapatkan_ikon(lok)          # Kustomisasi ikon (Penugasan 2)

        folium.Marker(
            location=koordinat,
            popup=folium.Popup(info_popup, max_width=300),
            tooltip=f"{lok.nama} ({type(lok).__name__})",
            icon=ikon
        ).add_to(peta)
        jumlah_marker += 1

    if lokasi_dilewati:
        pesan = f"Melewati marker: {', '.join(lokasi_dilewati)} (koordinat tidak valid)."
        print(f"  Peringatan: {pesan}")
        tulis_log(pesan, file_log)

    # 4. Simpan peta ke HTML dan tulis log
    try:
        peta.save(file_output)
        pesan_sukses = (f"Peta '{file_output}' berhasil dibuat "
                        f"dengan {jumlah_marker} marker.")
        print(f"\n✅ {pesan_sukses}")
        tulis_log(pesan_sukses, file_log)
    except Exception as e:
        pesan_error = f"ERROR menyimpan peta: {type(e).__name__} - {e}"
        print(f"\n❌ {pesan_error}")
        tulis_log(pesan_error, file_log)

    return peta   # Kembalikan objek peta agar bisa ditampilkan di Jupyter


# ── KODE UTAMA ────────────────────────────────────────────────────────────────
print("=" * 60)
print("  MINI PROJECT SIG – KOTA SEMARANG")
print("=" * 60)

# 1. Baca CSV
df_lokasi      = baca_data_lokasi('lokasi_semarang.csv')

# 2. Buat objek dari DataFrame
list_semua_lok = buat_objek_lokasi_dari_df(df_lokasi)

# 3. Buat peta (konfigurasi dibaca dari file, marker dikustomisasi)
peta_hasil = buat_peta_lokasi_folium(
    list_semua_lok,
    file_output='peta_semarang_final.html',
    file_config='config_peta.txt',
    file_log='proses_peta.log'
)

print("\nSelesai! File yang dihasilkan:")
print("  - peta_semarang_final.html  → buka di browser untuk lihat peta interaktif")
print("  - proses_peta.log           → file log proses")
```

3 Baca CSV & Buat Objek (dengan Kelas Baru)

```python
def baca_data_lokasi(nama_file: str):
    """Membaca file CSV ke dalam DataFrame Pandas."""
    try:
        df = pd.read_csv(nama_file)
        print(f"File '{nama_file}' berhasil dibaca. ({len(df)} baris)")
        return df
    except FileNotFoundError:
        print(f"ERROR: File '{nama_file}' tidak ditemukan!")
        return None
    except Exception as e:
        print(f"ERROR: {type(e).__name__} - {e}")
        return None


def buat_objek_lokasi_dari_df(dataframe: pd.DataFrame) -> list:
    """
    Mengiterasi DataFrame dan membuat list objek Lokasi
    berdasarkan kolom 'Tipe', termasuk kelas Museum dan KantorPemerintahan.
    """
    list_objek = []
    if dataframe is None or dataframe.empty:
        print("DataFrame kosong atau None.")
        return list_objek

    for index, row in dataframe.iterrows():
        nama      = row.get('Nama',      None)
        lat       = row.get('Latitude',  None)
        lon       = row.get('Longitude', None)
        tipe      = row.get('Tipe',      'Lainnya')
        deskripsi = row.get('Deskripsi', '')

        if nama is None or lat is None or lon is None:
            print(f"  Baris {index} dilewati: data tidak lengkap.")
            continue

        objek = None
        try:
            if 'Wisata' in tipe or tipe == 'Landmark':
                objek = TempatWisata(nama, lat, lon, tipe, deskripsi)
            elif tipe == 'Kuliner':
                objek = Kuliner(nama, lat, lon, deskripsi)
            elif 'Ibadah' in tipe:
                agama = "Islam" if "Islam" in tipe else "Umum"
                objek = TempatIbadah(nama, lat, lon, agama, deskripsi)
            # ── Tipe baru (Penugasan) ──────────────────────────────────
            elif tipe == 'Museum':
                objek = Museum(nama, lat, lon, koleksi_utama=deskripsi,
                               deskripsi=deskripsi)
            elif 'Kantor' in tipe:
                objek = KantorPemerintahan(nama, lat, lon,
                                           instansi=tipe,
                                           deskripsi=deskripsi)
            else:
                print(f"  Tipe '{tipe}' untuk '{nama}' tidak dikenali.")
        except Exception as e:
            print(f"  GAGAL membuat objek '{nama}': {e}")
            continue

        if objek:
            list_objek.append(objek)

    print(f"\nTotal {len(list_objek)} objek berhasil dibuat dari {len(dataframe)} baris data.")
    return list_objek


# ── Jalankan ─────────────────────────────────────────────────────────────────
df_lokasi       = baca_data_lokasi('lokasi_semarang.csv')
list_semua_lok  = buat_objek_lokasi_dari_df(df_lokasi)

print("\nDaftar objek yang dibuat:")
for i, lok in enumerate(list_semua_lok, 1):
    print(f"  {i}. {repr(lok)}")
```
